```python
dim_store['store_status'].values.unique()
active_dim_store=dim_store[dim_store['store_status']=='Active']
active_dim_store.head(5)
```

```python
dim_fact_salesstore=fact_daily_sales.merge(active_dim_store, on='store_id',how='left')
dim_fact_salesstore.head(4)
```

```python
sns.heatmap(region_detail_table.corr(),annot=True)
#setting annot to true label the corr inside the cell or each box
#From the chat u can see that there is a strong corrletion btw
#Loyalty_transction and stass_headcount(very strong)
#loylty_transaction and total_revenue
```

```python
sns.catplot(region_detail_table,kind='bar',x='region', y='daily_gross_profit_gbp')
plt.xticks(rotation=90)
plt.show()
```

```python
store_date_table = fact_daily_sales.merge(dim_date,on='date_id',how='left')
store_date_table.head()
```

```python
"""
we know that the footfall is the number of ppl the comes in the stroe and coming in those not mean u woukld buy somthing that is why we have total transaction
also we would create oppurtunity missed has we get what oppoutinty was missed by our store out of the foot fall also gte the conversion rate with this
"""

store_date_table['Total_missed_transaction']=store_date_table['footfall']-store_date_table['total_transactions']
store_date_table['customer_conversion_rate']=(store_date_table['total_transactions']/store_date_table['footfall'])*100

store_date_table[['Total_missed_transaction','customer_conversion_rate']] # checking if it works 
```

```python
year_total= store_date_table.groupby('year').agg({'total_revenue_gbp':'sum',
              'total_transactions':'sum',
              'loyalty_transactions':'sum',
              'loyalty_redemption_cost_gbp':'mean',
              'footfall':'sum',
              'daily_gross_profit_gbp':'sum',
              'total_staff_hours':'mean',
              'avg_dwell_time_minutes':'mean',
              'Total_missed_transaction': 'sum',
              'customer_conversion_rate':'mean'
             })

year_total
              
```

```python
year_total.plot(kind='bar',figsize=(10, 6))

plt.xlabel('Store ID')
plt.ylabel('Values')
plt.xticks(rotation=45)
plt.show()
```

```python
# checking the trend using the months of the year
month_total= store_date_table.groupby('month_name').agg({'total_revenue_gbp':'sum',
              'total_transactions':'sum',
              'loyalty_transactions':'sum',
              'loyalty_redemption_cost_gbp':'mean',
              'footfall':'sum',
              'daily_gross_profit_gbp':'sum',
              'total_staff_hours':'mean',
              'avg_dwell_time_minutes':'mean',
              'Total_missed_transaction': 'sum',
              'customer_conversion_rate':'mean'
             })

month_total
```
